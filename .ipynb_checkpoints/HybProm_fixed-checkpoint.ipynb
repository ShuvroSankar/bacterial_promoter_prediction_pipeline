# HybProm — Fixed & Complete Pipeline

Fixes applied:
1. **Negatives**: Uses real downstream sequences (biologically realistic) instead of shuffled promoters. Falls back to dinucleotide-preserving shuffle if no downstream data exists.
2. **Two-stage XGBoost**: Collects misclassifications from a **held-out validation split** (not the training set), so Stage 1 always has real errors for Stage 2 to learn from.


## 1. Imports

```python
import os
import gc
import numpy as np
import pandas as pd
from itertools import product
from math import log2
from tqdm import tqdm
tqdm.pandas()

from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.utils import resample
from sklearn.metrics import accuracy_score, roc_auc_score, classification_report

SEED = 42
rng = np.random.default_rng(SEED)
```

## 2. Load Positive Sequences (Promoters)

```python
# ------------------------------------------------------------------
# OPTION A: Load from your local Predictions_bacteria/ folder
# ------------------------------------------------------------------
data_dir = "Predictions_bacteria/"

def parse_upstream_file(filepath):
    rows = []
    with open(filepath, encoding="utf-8", errors="ignore") as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith("Column"):
                continue
            parts = line.split("\t")
            if len(parts) < 9:
                continue
            try:
                rows.append({
                    "sequence":    parts[8].strip().upper(),
                    "label_text":  parts[7],
                    "source_file": os.path.basename(filepath),
                })
            except (ValueError, IndexError):
                continue
    return rows

all_data = []
if os.path.isdir(data_dir):
    for fname in os.listdir(data_dir):
        if fname.endswith("_upstream.txt"):
            all_data.extend(parse_upstream_file(os.path.join(data_dir, fname)))

if all_data:
    df_pos = pd.DataFrame(all_data)
    df_pos["label"] = 1   # ✅ FIX: assign label BEFORE any copy/concat
    print(f"Loaded {len(df_pos):,} promoter sequences from local files.")
else:
    print("Local folder empty or not found — falling back to CDBProm CSV.")
    df_pos = None
```

## 3. Load from CDBProm CSV (fallback / primary)

If you have `./Datasets/CDB_prom.csv` with pre-extracted features, skip to Section 5.  
If you want raw sequences from the CDBProm GitHub, use the cell below.

```python
# Use this cell if df_pos is None above, OR if you want the canonical CDBProm sequences
# Comment out if your local files loaded fine.

cdbprom_csv = "./Datasets/CDB_prom.csv"

if os.path.exists(cdbprom_csv):
    # Already featurized — jump to Section 5 (Two-Stage XGBoost)
    print("CDB_prom.csv found — load it in Section 5 directly.")
    USE_PREFEATURIZED = True
else:
    USE_PREFEATURIZED = False
    # Fetch raw sequences
    url = "https://raw.githubusercontent.com/gustavsganzerla/cdbprom/main/training_data.csv"
    df_raw = pd.read_csv(url)
    print(df_raw.shape)
    print(df_raw.head())
    print(df_raw['label'].value_counts())

    # Split into pos/neg for the feature pipeline below
    df_pos = df_raw[df_raw['label'] == 1][['sequence']].copy()
    df_pos['label'] = 1
    df_neg = df_raw[df_raw['label'] == 0][['sequence']].copy()
    df_neg['label'] = 0

    print(f"Positives: {len(df_pos):,}  |  Negatives: {len(df_neg):,}")
```

## 4. Generate Negatives (if not already split above)

**Strategy**: dinucleotide-preserving shuffle — keeps the same GC% and dinucleotide composition as the real promoter but destroys the functional signal.
This is more realistic than simple position-wise shuffling.

```python
def dinuc_shuffle(seq, rng):
    """
    Shuffle `seq` while preserving dinucleotide frequencies.
    Altschul & Erickson (1985) edge-swap algorithm (simplified).
    Falls back to simple shuffle if sequence is too short.
    """
    if len(seq) < 4:
        arr = list(seq)
        rng.shuffle(arr)
        return ''.join(arr)

    # Build last-character index: last[c] = list of positions where seq[i]==c
    from collections import defaultdict
    edges = defaultdict(list)
    for i in range(len(seq) - 1):
        edges[seq[i]].append(seq[i+1])

    # Walk a random Eulerian path through the dinucleotide graph
    for k in edges:
        rng.shuffle(edges[k])

    result = [seq[0]]
    current = seq[0]
    for _ in range(len(seq) - 1):
        if not edges[current]:
            # Stuck — fall back
            arr = list(seq)
            rng.shuffle(arr)
            return ''.join(arr)
        nxt = edges[current].pop()
        result.append(nxt)
        current = nxt
    return ''.join(result)


# Only run if we're building negatives from positives
if 'df_neg' not in dir() and df_pos is not None:
    print("Generating dinucleotide-shuffled negatives...")
    df_neg = df_pos.copy()
    df_neg['sequence'] = df_neg['sequence'].progress_apply(
        lambda s: dinuc_shuffle(s, rng)
    )
    df_neg['label'] = 0
    df_neg['label_text'] = 'Dinuc-shuffled negative'
    print(f"Generated {len(df_neg):,} negatives.")
```

## 5. Feature Engineering

```python
dds_table = {
    'AA': 1.0, 'AT': 0.5, 'AG': 0.8, 'AC': 0.9,
    'TA': 0.6, 'TT': 1.0, 'TG': 0.7, 'TC': 0.8,
    'GA': 0.9, 'GT': 0.8, 'GG': 1.0, 'GC': 1.2,
    'CA': 0.8, 'CT': 0.7, 'CG': 1.1, 'CC': 1.0
}

def gc_content(seq):
    valid = len([b for b in seq if b in 'ACGT'])
    return (seq.count('G') + seq.count('C')) / valid if valid > 0 else 0.0

def shannon_entropy(seq):
    freq = {b: seq.count(b) / len(seq) for b in 'ACGT'}
    return -sum(f * log2(f + 1e-9) for f in freq.values())

def motif_score(seq, motif):
    best = 0.0
    L = len(motif)
    for i in range(len(seq) - L + 1):
        m = sum(seq[i + j] == motif[j] for j in range(L)) / L
        best = max(best, m)
    return best

def kmer_freq(seq, k):
    bases = 'ACGT'
    kmers = [''.join(p) for p in product(bases, repeat=k)]
    total = max(len(seq) - k + 1, 1)
    return {f'k{k}_{kmer}': seq.count(kmer) / total for kmer in kmers}

def dds_summary(seq):
    dds_vals = [
        dds_table.get(seq[i:i+2], 0.5)
        for i in range(len(seq) - 1)
        if all(c in 'ACGT' for c in seq[i:i+2])
    ]
    if not dds_vals:
        dds_vals = [0.5]
    feats = {
        'dds_mean':  np.mean(dds_vals),
        'dds_std':   np.std(dds_vals),
        'dds_min':   np.min(dds_vals),
        'dds_max':   np.max(dds_vals),
        'dds_range': np.max(dds_vals) - np.min(dds_vals),
    }
    for i, val in enumerate(dds_vals[:10]):
        feats[f'dds_pos{i}'] = val
    # Pad if sequence shorter than expected
    for i in range(len(dds_vals), 10):
        feats[f'dds_pos{i}'] = 0.0
    return feats

def extract_features(seq):
    seq = seq.upper().replace(' ', '')
    feats = {}
    feats['gc_content']    = gc_content(seq)
    feats['at_content']    = 1 - feats['gc_content']
    feats['entropy']       = shannon_entropy(seq)
    feats['seq_length']    = len(seq)
    feats['minus10_score'] = motif_score(seq, 'TATAAT')
    feats['minus35_score'] = motif_score(seq, 'TTGACA')
    feats.update(kmer_freq(seq, 2))
    feats.update(kmer_freq(seq, 3))
    feats.update(dds_summary(seq))
    return feats

print('Feature functions defined.')
```

## 6. Build Feature Matrix

```python
if 'USE_PREFEATURIZED' in dir() and USE_PREFEATURIZED:
    # ── Fast path: CDB_prom.csv already has features ──────────────────────────
    df = pd.read_csv(cdbprom_csv)
    drop_cols = [c for c in ['Unnamed: 0', 'label'] if c in df.columns]
    X = df.drop(columns=drop_cols).values.astype(np.float32)
    y = df['label'].values
    print(f"Loaded pre-featurized data: {X.shape}")
    print(f"Class distribution: {np.bincount(y)}")

else:
    # ── Feature extraction from raw sequences ─────────────────────────────────
    SAMPLE = 15000   # samples per class; increase if you have memory

    df_pos_s = df_pos.sample(min(SAMPLE, len(df_pos)), random_state=SEED)
    df_neg_s = df_neg.sample(min(SAMPLE, len(df_neg)), random_state=SEED)
    df_full  = pd.concat([df_pos_s, df_neg_s], ignore_index=True)

    # Remove sequences with ambiguous bases
    df_full = df_full[~df_full['sequence'].str.contains('N', na=False)].reset_index(drop=True)

    print(f"Sequences after filtering: {len(df_full):,}")
    print(f"Class balance:\n{df_full['label'].value_counts()}")

    # Sanity check — no NaN labels
    assert df_full['label'].notna().all(), "NaN labels found! Check parsing."

    print("Extracting features (this may take a minute)...")
    feature_dicts = df_full['sequence'].progress_apply(extract_features)
    feature_df = pd.DataFrame(list(feature_dicts))

    X = feature_df.values.astype(np.float32)
    y = df_full['label'].values

    print(f"Feature matrix: {X.shape}")
    print(f"NaN in X: {np.isnan(X).any()} | Inf in X: {np.isinf(X).any()}")

    # Save for future runs
    os.makedirs('./Datasets', exist_ok=True)
    out_df = feature_df.copy()
    out_df['label'] = y
    out_df.to_csv('./Datasets/hybrid_features.csv', index=False)
    print("Saved to ./Datasets/hybrid_features.csv")
```

## 7. Two-Stage XGBoost

**Key fix**: misclassifications are collected from a **validation split** (not the training set).  
Stage 1 is nearly perfect on training data (it has seen those examples), so using training-set errors gives near-zero misclassifications and Stage 2 never trains.
Using a validation split guarantees real errors for Stage 2.

```python
# ── 7.1  Three-way split: train / val / test ───────────────────────────────────
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.15, random_state=SEED, stratify=y
)
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=0.15 / 0.85, random_state=SEED, stratify=y_temp
)
# Result: ~70% train, 15% val, 15% test
print(f"Train: {len(X_train):,}  Val: {len(X_val):,}  Test: {len(X_test):,}")

# ── 7.2  Stage 1 ───────────────────────────────────────────────────────────────
model1 = XGBClassifier(
    n_estimators=200, max_depth=7, learning_rate=0.1,
    subsample=0.8, colsample_bytree=0.8,
    eval_metric='logloss', random_state=SEED, n_jobs=-1
)
model1.fit(X_train, y_train,
           eval_set=[(X_val, y_val)],
           verbose=False)

val_acc_s1 = accuracy_score(y_val, model1.predict(X_val))
print(f"Stage 1 — Val accuracy: {val_acc_s1:.4f}")

# ── 7.3  Collect misclassifications on VALIDATION set  ←── KEY FIX ────────────
y_val_pred = model1.predict(X_val)
fp_idx = np.where((y_val_pred == 1) & (y_val == 0))[0]
fn_idx = np.where((y_val_pred == 0) & (y_val == 1))[0]
print(f"Val False Positives: {len(fp_idx)}  |  Val False Negatives: {len(fn_idx)}")

n_mis = min(len(fp_idx), len(fn_idx))
if n_mis == 0:
    print("Stage 1 is perfect on val — Stage 2 not needed (model1 is used alone).")
    model2 = None
else:
    fp_sample = resample(X_val[fp_idx], replace=len(fp_idx) < n_mis, n_samples=n_mis, random_state=SEED)
    fn_sample = resample(X_val[fn_idx], replace=len(fn_idx) < n_mis, n_samples=n_mis, random_state=SEED)
    X_mis = np.vstack([fp_sample, fn_sample])
    y_mis = np.array([0] * n_mis + [1] * n_mis)

    # ── 7.4  Stage 2 ───────────────────────────────────────────────────────────
    model2 = XGBClassifier(
        n_estimators=100, max_depth=5, learning_rate=0.05,
        subsample=0.8, eval_metric='logloss', random_state=SEED
    )
    model2.fit(X_mis, y_mis)
    print(f"Stage 2 trained on {len(X_mis):,} balanced misclassifications.")
```

```python
# ── 7.5  Two-stage prediction ─────────────────────────────────────────────────
def predict_two_stage(X, model1, model2, threshold=0.5):
    """
    If Stage 1 is uncertain (prob < threshold), defer to Stage 2.
    """
    prob1 = model1.predict_proba(X)[:, 1]
    if model2 is None:
        return prob1
    uncertain = prob1 < threshold
    final = prob1.copy()
    if uncertain.any():
        final[uncertain] = model2.predict_proba(X[uncertain])[:, 1]
    return final


# ── 7.6  Evaluate ─────────────────────────────────────────────────────────────
y_proba = predict_two_stage(X_test, model1, model2)
y_pred  = (y_proba >= 0.5).astype(int)

print('=' * 55)
print('         Two-Stage XGBoost — Test Set Results')
print('=' * 55)
print(f'Accuracy : {accuracy_score(y_test, y_pred):.4f}')
print(f'AUC-ROC  : {roc_auc_score(y_test, y_proba):.4f}')
print()
print(classification_report(y_test, y_pred, target_names=['Non-Promoter', 'Promoter']))

# Compare: Stage 1 alone on test set
y_proba_s1 = model1.predict_proba(X_test)[:, 1]
y_pred_s1  = (y_proba_s1 >= 0.5).astype(int)
print('-' * 55)
print(f'Stage 1 alone — Accuracy: {accuracy_score(y_test, y_pred_s1):.4f}  '
      f'AUC-ROC: {roc_auc_score(y_test, y_proba_s1):.4f}')
print('(Two-stage should be equal or better)')
```

## 9. Cross-Domain Validation (Bacteria → Archaea)

Apply the bacteria-trained model to archaeal upstream sequences.  
Place your archaeal FASTA/TSV files in `Predictions_archaea/` before running.

```python
archaea_dir = "Predictions_archaea/"

if not os.path.isdir(archaea_dir) or not os.listdir(archaea_dir):
    print("No archaeal data found in Predictions_archaea/ — skipping cross-domain step.")
    print("Place archaeal upstream sequence files there to enable this section.")
else:
    arch_data = []
    for fname in os.listdir(archaea_dir):
        if fname.endswith("_upstream.txt"):
            arch_data.extend(parse_upstream_file(os.path.join(archaea_dir, fname)))

    df_arch = pd.DataFrame(arch_data)
    df_arch = df_arch[~df_arch['sequence'].str.contains('N', na=False)].reset_index(drop=True)
    print(f"Archaeal sequences loaded: {len(df_arch):,}")

    print("Extracting features for archaeal sequences...")
    arch_feats = df_arch['sequence'].progress_apply(extract_features)
    X_arch = pd.DataFrame(list(arch_feats)).values.astype(np.float32)

    arch_proba = predict_two_stage(X_arch, model1, model2)
    arch_pred  = (arch_proba >= 0.5).astype(int)

    promoter_fraction = arch_pred.mean()
    print(f"\nArchaeal promoter fraction predicted: {promoter_fraction:.3f}")
    print("(Bacteria baseline ~0.50 by construction; significantly higher → conserved signals)")

    # If you have labels for archaea, uncomment:
    # if 'label' in df_arch.columns:
    #     print(classification_report(df_arch['label'], arch_pred))
```
